# **Vectorized** Math *fn* in `NumPy`

#### Vectorized Math Functions in NumPy allows us to **perform mathematical operation** on an entire array without using the `for loops`


|**Function**|**description**|
|---|---|
|`np.sqrt()`|*square root*|
|`np.square()`|*square*|
|`np.abs()`|*absolute*|
|`np.exp()`|*exponent*|
|`np.log()`|*logarithmic (Natural)*|
|`np.log10()`|*logarithmic (base-10)*|
|`np.sin()`|*Sine*|
|`np.cos()`|*Cosine*|
|`np.tan()`|*Tangent*|
|`np.ceil()`|*round-up*|
|`np.floor()`|*round-down*|
|`np.round()`|*round value*|


```python
import numpy as np
```

```python
array = np.array([1, 4, 9, 16])
```

#### 1. `np.sqrt()`

```python
print(np.sqrt(array))
```

2. `np.square()`

```python
print(np.square(array))
```

```python
array = np.array([-2.5, 3.2, -4.7, 5.8])
```

#### 3. `np.abs()`

```python
print(np.abs(array))
```

#### 4. `np.ceil()`

```python
print(np.ceil(array))
```

#### 5. `np.floor`

```python
print(np.floor(array))
```

#### 6. `np.floor()`

```python
print(np.floor(array))
```
