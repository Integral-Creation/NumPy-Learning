# Array ***Arithmetic*** in `NumPy`

#### `NumPy` allows you to perform arithmetic operations element-by-element on arrays.

```python
import numpy as np
```

```python
array = np.array([1,2,3,4,5])
```

## `Scalar` Arithmetic

#### Performing the array arithmetic operation between a NumPy array and a single value(scalar).

```python
# Addition
print(array + 5)
```

```python
# Subtraction
print(array - 5)
```

```python
# Multiplication
print(array * 5)
```

```python
# Division 
print(array / 5)
```

```python
# Floor division 
print(array // 2)
```

```python
# Power 
print(array ** 2)
```
