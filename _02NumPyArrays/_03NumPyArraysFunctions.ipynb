## **`NumPy` functions**

```python
import numpy as np
```

### **`NumPy` array Creations**

#### `.zero()` : create an array filled with *zeros*

```python
array = np.zeros(5)
print(array)
```

#### creating `2D` array filled with zeros

```python
array = np.zeros((2,3))
print(array)
```

#### creating `3D` array filled with zeros

```python
array = np.zeros((3,3,3))
print(array)
```

#### `.ones()` : create an array filled with ones

```python
array = np.ones((5))
print(array)
```

#### `.full()` create an array filled with the *value parameter*

#### creating `1D` array using `.full()`

```python
array = np.full(5,10)
print(array)
```

#### creating `2D` array using `.fill()`

```python
array = np.full([2,2],10)
print(array)
```

#### `.arange()` creates a array within the given interval

```python
array = np.arange(5,10)
print(array)
```

#### `.linspace()` creates an arrays of evenly spaced values between the given values

```python
array = np.linspace(0,1,5)
print(array)
```

#### `.random.rand()` creates an array of random values between 0 and 1

```python
array = np.random.rand(5)
print(array)
```

#### `.random.randint` creates an arrays of random integer values within the specified range

#### creating a `1D` array using `.random.randint()`

```python
array = np.random.randint(0,5, size= 5)
print(array)
```

#### creating a `2D` array using `.random.randint()`

```python
array = np.random.randint(0, 10, size= (2, 3))
print(array)
```

#### creating a `3D` array using `.random.randint()`

```python
array = np.random.randint(4,16,size=(3,3,4))
print(array)
```

```python
array = np.arange(0,10,3)
print(array)
```

```python
array = np.arange(4).reshape(4)
print(array)
```
