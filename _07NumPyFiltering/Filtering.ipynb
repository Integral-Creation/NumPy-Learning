# **Filtering** Data in **`NumPy`**

#### `Filtering` in NumPy means selecting specific elements from an array based on a condition. It is commonly performed using Boolean indexing.

```python
import numpy as np
```

## 1. `Filtering` Using a Condition


```python
arr = np.array([2,3,4,5,12,15,16])
```

```python
result = arr[arr > 5] # value greater than 5
print(result)
```

```python
result = arr[arr %2 == 0] # even values
print(result)
```

## 2. `Filtering` Using Multiple Conditions

```python
result = arr[(arr > 5) & (arr %2 == 0)] # even values greater than 5
print(result)
```

## 3. `Filtering` with **`np.where()`**

#### `np.where()` function returns the indices of the elements that satisfy the condition.

```python
arr = np.array([10,20,30,40,50])

result = arr[np.where(arr > 25)]
print(result)
```

#### `np.where()` can be used to replace values conditionally

```python
result = np.where(arr > 25, arr, 0)
print(result)
```
